```python
import pandas as pd  

df = pd.read_csv("data/raw/scb_inkomst_00_03.csv", skiprows=2, encoding="latin")

if "inkomstklass" in df.columns:
    df = df.drop(columns=["inkomstklass"])

print(df.head())
```

```python
# omvandlar åren till en enda kolumn
df_melted = pd.melt(
    df,
    id_vars=["region", "utbildningsnivå", "kön", "ålder"], # kolumner som ska behållas
    value_vars=["2000", "2001", "2002", "2003"], # kolumner som ska omvandlas till rader
    var_name="year",
    value_name="inkomst_tkr"
)

print(df_melted.head())
```
